3. \
Justifique analiticamente si es posible aplicar el metodo de Newton a la funcion $f(x) = cos(x)-2x+3$, en caso afirmativo determine la raiz de $f(x)$ en el intervalo $[1, 2]$. Muestre una tabla en la que se muestre la iteracion, la aproximacion de la raiz y la evaluacion de la raiz en la funcion.

Sea $f(x) = cos(x)-2x+3$, queremos garantizar que converge a una raiz en el intervalo $I = [1, 2]$, entonces analizaremos si es diferenciable.\
Derivando tenemos $$f'(x)=-sen(x)-2$$
luego analizamos si existe una raiz en el intervalo $I$\
asi\
$$f(1) = cos(1)-2+3 = 1.5403 > 0$$
$$f(2) = cos(2)-4+3 = -1.4161 < 0$$
dado que hay cambio de signo podemos afirmar por el TVI que existe al menos una raiz en el intervalo $[1, 2]$

```python
import math
def f(x):
  return math.cos(x)-2*x+3

def fp(x):
  return -1*math.sin(x)-2
i = 10 #int(input("Ingrese el numero de iteraciones a realizar: "))
p0 = 1.5 #float(input("Ingrese el punto inicial: "))
print("="*50)
print(f"{'n':^4} | {'p_n':^8} | {'Aprox':^10} | {'Evalua en f(x)':^10}")
print("="*50)
for n in range(i):
  pm = p0 - (f(p0)/fp(p0))
  ev = f(pm)
  print(f"{n:^4} | {p0:^8.3f} | {pm:^10.6f} | {ev:^10.8f}")
  p0 = pm
```

4. \
Sea $f(x) = cos(x) -3sen(x) - x^3$, queremos garantizar que converge en el intervalo $[0, 1]$, obs que $f(x)$ es continua y diferenciable, entonces\
Derivando tenemos
$$f'(x) = -sen(x) -3cos(x) -3x^2$$
luego analizamos si existe una raiz en el intervalo $[0,1]$\
asi
$$f(0) = cos(0) -3sen(0) - (0)^3 = 1 > 0$$
$$f(1) = cos(1) -3sen(1)-(1)^3 = -2.9841 < 0$$
dado que hay un cambio de signo podemos afirmar por el TVI que existe al menos una raiz en el intervalo $[0,1]$.

```python
import math
def f(x):
  return math.cos(x)-3*math.sin(x)-x**3

def fp(x):
  return -math.sin(x)-3*math.cos(x)-3*x**2
i = 10 #numero de iteraciones
p0 = 0.5 #punto inicial
print("="*50)
print(f"{'n':^4} | {'p_n':^8} | {'Aprox':^15} | {'Evalua en f(x)':^15}")
print("="*50)

for n in range(i):
  pm = p0 - (f(p0)/fp(p0))
  ev = f(pm)
  print(f"{n:^4} | {p0:^8.3f} | {pm:^15.6f} | {ev:^15.6f}")
  p0 = pm
```

5. \
Sea $f(x) = x^3 + x^2 - 1$, queremos garantizar que converge en el intervalo $[0, 1]$, obs que $f(x)$ es continua y diferenciable, entonces\
Derivando tenemos
$$f'(x) = 3x^2 + 2x$$
luego analizamos si existe una raiz en el intervalo $[0,1]$\
asi
$$f(0) = (0)^3 + (0)^2 - 1 = -1 < 0$$
$$f(1) = (1)^3 + (1)^2 - 1 = 1 > 0$$
dado que hay un cambio de signo podemos afirmar por el TVI que existe al menos una raiz en el intervalo $[0,1]$.

```python
def f(x):
  return x**3 + x**2 -1

def fp(x):
  return 3*x**2 + 2*x

i = 10
p0 = 0.5
print("="*50)
print(f"{'n':^4} | {'p_n':^8} | {'Aprox':^15} | {'Evalua en f(x)':^15}")
print("="*50)

for n in range(i):
  pm = p0 - (f(p0)/fp(p0))
  ev = f(pm)
  print(f"{n:^4} | {p0:^8.3f} | {pm:^15.6f} | {ev:^15.6f}")
  p0 = pm
```
