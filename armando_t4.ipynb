1.- Hacer un programa para el metodo de punto fijo.

```python
fun = input("Ingrese la funcion g(x): ")
i = int(input("Ingrese el numero de iteraciones a realizar: "))
p0 = int(input("Ingrese el punto incial: "))

def g(x):
    reemplazo = fun.replace('x', str(x))
    evaluacion = eval(reemplazo)
    return evaluacion
for n in range(i):
  pm = g(p0)
  p0 = pm
  print(f"p_{n+1} = {pm}")
```

2. a)\
De un teorema visto en clase sabemos que:
$$p_m = g(p_{m-1})$$ con $m \geq 1$, entoces hacemos $g(x)=\frac{2}{x}$ con $p_0 = 1$

```python
def g(x):
    return 2/x

p0 = 1
for n in range(10):
    pm = g(p0)
    p0 = pm
    print(f"p_{n+1} = {p0}")
print("Observemos que la sucesion g(x) oscila entre 1 y 2 por lo tanto la sucesion diverge\n")
```

2. b)\
Sea $p_m = g(p_{m-1})$ entoces hacemos $g(x)=\frac{1}{2}(x+cos(x))$ con $p_0 = 2$ con 15 iteraciones

```python
import math
def g(x):
    return (1/2)*(x + math.cos(x))

p0 = 2
for n in range(15):
    pm = g(p0)
    p0 = pm
    print(f"p_{n+1} = {p0}")
print("\n Por lo tanto la sucesion converge \n")
```

$$\Longrightarrow x_1 = \frac{-3 +\sqrt{13}}{2}$$
obs que $x_0 \notin [-1, 1]$ por lo tanto el unico punto fijo es $x_1$

5. \
Sea $f(x)=x^3+4x^2-10$, despejando\
$x^3+4x^2 = 10$\
$\Longrightarrow x^2(x+4) = 10$\
$\Longrightarrow x = \sqrt{\frac{10}{x+4}}$\
Luego sea $g(x)= \sqrt{\frac{10}{x+4}}$ derivando tenemos
$$\frac{1}{2\sqrt{\frac{10}{x+4}}}-\frac{10}{(x+4)^2} = -\frac{5}{\sqrt{\frac{10}{x+4}}(x+4)^2} = g'(x)$$\
obs si $|g'(x)|< 1$
$\Longrightarrow |g'(x)| = \frac{5}{\sqrt{\frac{10}{x+4}}(x+4)^2}$\
para $x=1$\
$\frac{5}{25\sqrt{2}} = 0.141$\
para $x=2$\
$\frac{5}{36\sqrt{\frac{10}{6}}} = 0.117$\
como $|g'(x)|< 1$ en $[1, 2]$ podemos garantizar la convergencia.

```python
import math
# funcion obtenida al despejar x
def g(x):
    return math.sqrt(10/(x+4))
#funcion dada en el ununciado
def ev(x):
    return x**3+4*x**2-10

i = 10 #int(input("Ingrese el numero de iteraciones a realizar: "))
print("="*55)
print(f"| {'n':^4} | {'p0':^8} | {'Aproximacion':^15} | {'Evalua en f(x)':^15} |")
print("="*55)
p0 = 1.5
for n in range(i):
    pm = g(p0)# obtiene la aproximacion
    p0 = pm
    e = ev(pm)# evalua en la funcion dada
    print(f"| {n+1:^4} | {p0:^8.5f} | {pm:^15.10f} | {e:^15.10f} |")
print("="*55)
```
